## Introduction

Run:ai (now part of NVIDIA) provides a **commercial GPU scheduling and orchestration platform** for Kubernetes.

### What is it?

- A platform that integrates with Kubernetes to manage **GPU resources**, including fractional GPUs and time-slicing.  
- Offers **multi-tenant GPU sharing, quota management, and observability** for AI/ML workloads.  
- Includes a scheduler and management plane that sits alongside your Kubernetes cluster.

### Why use it?

Key benefits of using Run:ai:

- **Improved GPU utilization** via fractional GPU allocation and time-slicing.  
- **Enterprise features**: RBAC, projects, quotas, UI dashboards, and reporting.  
- **Support and integrations** for enterprise environments.

### When to use it?

Run:ai is particularly useful when:

- You operate **large GPU clusters** and need to maximize utilization.  
- You want a **managed / vendor-supported** solution for GPU scheduling.  
- You need per-team GPU quotas, fair-sharing, and usage tracking.

## Key Features

### Core Capabilities of Run:ai

| Feature | Description | Benefit |
|--------|-------------|---------|
| **GPU fractions & time-slicing** | Share a single GPU among multiple workloads. | Higher utilization and more users per GPU. |
| **Project-based quotas** | Allocate GPU/CPU quotas to teams or projects. | Strong multi-tenant resource governance. |
| **Scheduling policies** | Control placement and fairness across workloads. | SLA adherence and predictable performance. |
| **Dashboards & reporting** | Visualize GPU usage, queues, and historical consumption. | Capacity planning and cost management. |
| **Integration with Kubernetes** | Works with existing clusters and workloads. | Incremental adoption without rewriting apps. |

## Best Practices

1. **Work with platform teams**  
   - Understand project quotas, GPU pools, and policies.  

2. **Right-size GPU requests**  
   - Use fractions/time-slicing where appropriate; avoid over-requesting full GPUs for small experiments.

3. **Leverage dashboards**  
   - Use Run:ai’s UI to monitor your jobs’ GPU usage and tune workloads accordingly.

4. **Group workloads logically**  
   - Use projects and labels to align workloads with teams or initiatives.

## Common Pitfalls

1. **Misunderstanding GPU sharing semantics**  
   - Symptom: Performance variability when sharing GPUs.  
   - Fix: Align expectations about time-slicing/fractions and test workloads under shared conditions.

2. **Quota surprises**  
   - Symptom: Jobs failing to start due to quota limits.  
   - Fix: Coordinate with platform teams on quotas and resource planning.

3. **Vendor lock-in concerns**  
   - Symptom: Worry about dependence on a proprietary stack.  
   - Fix: Evaluate integration points and migration paths as part of adoption decisions.

## Performance Optimization

- **Profile workloads under shared vs exclusive GPUs** to understand trade-offs.
