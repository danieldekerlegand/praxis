## Introduction

Apache YuniKorn is a **universal resource scheduler** for container orchestrators such as Kubernetes.

### What is it?

- A pluggable scheduler designed for **multi-tenant, large-scale clusters**.  
- Provides advanced scheduling policies (bin-packing, gang scheduling, hierarchical queues).  
- Runs as a scheduler component in Kubernetes clusters.

### Why use it?

Key benefits of using YuniKorn:

- **Fine-grained resource sharing** across teams and workloads.  
- **Advanced placement policies** (e.g., bin-packing to reduce cost).  
- **Centralized view of resource usage** with a UI/dashboard.

### When to use it?

YuniKorn is particularly useful when:

- You operate **large, multi-tenant clusters** with many concurrent workloads.  
- You need advanced scheduling policies to reduce fragmentation and improve utilization.  
- You want a general-purpose scheduler that can handle **AI/ML, big data, and generic workloads**.

## Key Features

### Core Capabilities of YuniKorn

| Feature | Description | Benefit |
|--------|-------------|---------|
| **Hierarchical queues** | Organize workloads by org, team, or project. | Strong multi-tenant isolation and fairness. |
| **Bin-packing** | Pack workloads tightly on nodes to minimize cost. | Better resource utilization in cloud environments. |
| **Gang scheduling** | Run sets of pods together when all resources are available. | Useful for distributed AI/ML and big data jobs. |
| **Centralized UI** | Web UI to view queues, usage, and running workloads. | Improved visibility and operations. |
| **K8s scheduler plugin** | Integrate with Kubernetes via scheduler plugin. | Use YuniKorn as the main scheduler for pods. |

## Architecture Overview

YuniKorn runs as a scheduler that integrates with Kubernetes.

```text
+---------------------------+
|   Kubernetes API Server   |
+---------------------------+
          ^         |
          |         |  Pod/Node state
          |         v
+----------+----------------+
|   YuniKorn Scheduler      |
|  • Queues & policies      |
|  • Scheduling decisions   |
+----------+----------------+
           |
           v
+---------------------------+
|    Kubernetes Nodes       |
+---------------------------+
```

Pods are submitted as usual; YuniKorn handles placement based on its internal policies and queues.

## Basic Usage

As an ML practitioner, you typically:

- Target specific **queues** via annotations or labels defined by platform teams.  
- Request resources (CPU, memory, GPU) as usual in pod specs.

The actual configuration of queues and policies is managed centrally.

````python
# Example: a pod targeting a YuniKorn queue via annotation (conceptual)

yunikorn_pod_yaml = """
apiVersion: v1
kind: Pod
metadata:
  name: ml-train-pod
  annotations:
    yunikorn.apache.org/queue: root.ai-team
spec:
  containers:
  - name: trainer
    image: your-registry/trainer:latest
    resources:
      requests:
        cpu: "4"
        memory: "16Gi"
        nvidia.com/gpu: 1
  restartPolicy: Never
"""

print(yunikorn_pod_yaml)
````

## Best Practices

1. **Define a clear queue hierarchy**  
   - Map org structure into queues (e.g., root.dev, root.prod, root.research).

2. **Align policies with business priorities**  
   - Use capacity and priority settings that reflect real organizational needs.

3. **Use bin-packing where appropriate**  
   - Reduce cloud costs by packing workloads tightly, especially for bursty environments.

4. **Expose dashboards to users**  
   - Give ML engineers visibility into queue usage and bottlenecks.

## Common Pitfalls

1. **Complex queue hierarchies**  
   - Symptom: Hard to understand scheduling behavior.  
   - Fix: Keep queue hierarchy manageable and well-documented.

2. **Misconfigured capacities**  
   - Symptom: Some queues always full, others underused.  
   - Fix: Adjust capacity allocations and monitor usage regularly.

3. **Insufficient observability**  
   - Symptom: Difficulty explaining why some workloads are delayed.  
   - Fix: Use YuniKorn’s UI and metrics to improve transparency.

## Performance Optimization

- **Tune bin-packing vs. spreading** policies based on workload characteristics.  
- **Use separate node pools** for different workload types (CPU vs GPU, spot vs on-demand).
